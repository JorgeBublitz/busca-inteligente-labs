# Detecção de duplicatas
Calcula a matriz de similaridade semântica e lista pares acima do limiar escolhido.

```python
import json
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer

limiar = 0.85
dados = pd.DataFrame(json.load(open('manifestacoes.json', encoding='utf-8')))
modelo = SentenceTransformer('paraphrase-multilingual-MiniLM-L12-v2')
vetores = modelo.encode(dados.texto.tolist(), normalize_embeddings=True)
matriz = cosine_similarity(vetores)
sns.heatmap(matriz, cmap='Blues')
plt.title('Matriz de similaridade semântica')
plt.show()
pares = []
for i in range(len(dados)):
    for j in range(i + 1, len(dados)):
        if matriz[i,j] >= limiar:
            pares.append({'origem': dados.id[i], 'destino': dados.id[j], 'similaridade': matriz[i,j]})
pd.DataFrame(pares).sort_values('similaridade', ascending=False).round(3)
```
